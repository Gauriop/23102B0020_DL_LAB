```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import os
import cv2
print("ok")
```

```python
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

```python
(X_train,y_train),(X_test,y_test)=keras.datasets.mnist.load_data()
```

```python
X_train.shape
```

```python
X_train[0]
```

```python
X_test.shape
```

```python
X_test[0]
```

```python
plt.imshow(X_train[2])
```

```python
X_train=X_train/255
X_test=X_test/255
```

```python
X_train[0]
```

```python
model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),  # 28x28 image → 784 numbers
    keras.layers.Dense(128, activation='relu'),  # hidden layer 1
    keras.layers.Dense(128, activation='relu'),  # hidden layer 2
    keras.layers.Dense(10, activation='softmax') # output: 10 digits (0-9)
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history=model.fit(X_train,y_train,epochs=10,validation_split=0.2,verbose=1)
```

```python
y_prob=model.predict(X_test)
```

```python
y_prob
```

```python
y_pred=y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
```

```python
model.save('handwritten.keras')
print("Model saved!")
```

```python
model = tf.keras.models.load_model('handwritten.keras')

loss, accuracy = model.evaluate(X_test, y_test, verbose=0)

print("Loss    :", loss)
print("Accuracy:", accuracy)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(history.history['accuracy'],     label='Train accuracy')
ax1.plot(history.history['val_accuracy'], label='Val accuracy')
ax1.set_title('Accuracy over epochs')
ax1.set_xlabel('Epoch')
ax1.legend()

ax2.plot(history.history['loss'],     label='Train loss')
ax2.plot(history.history['val_loss'], label='Val loss')
ax2.set_title('Loss over epochs')
ax2.set_xlabel('Epoch')
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
predictions = model.predict(X_test)

fig, axes = plt.subplots(2, 5, figsize=(12, 5))
axes = axes.flatten()

for i, ax in enumerate(axes):
    predicted = np.argmax(predictions[i])
    actual    = y_test[i]
    color     = 'green' if predicted == actual else 'red'

    ax.imshow(X_test[i], cmap='gray')
    ax.set_title(f'Pred: {predicted}  Actual: {actual}', color=color, fontsize=9)
    ax.axis('off')

plt.suptitle('Green = correct prediction, Red = wrong', fontsize=11)
plt.tight_layout()
plt.show()
```
